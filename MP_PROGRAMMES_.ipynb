Determine nth roots of a complex number and represent it in Cartesian form and polar form

```python
import cmath
import math
import matplotlib.pyplot as plt

z = 8 + 8j
n = 3

r = abs(z)
th = cmath.phase(z)

rk = r**(1/n)

x = []
y = []

for k in range(n):
    thk = (th + 2 * math.pi * k) / n
    c = cmath.rect(rk, thk)

    print("cartesian:", c)
    print("polar:", rk, thk)
    print("")

    x.append(c.real)
    y.append(c.imag)

plt.plot(x, y, 'o')
plt.show()
```

Transformation of complex number as 2D vectors (Rotation)

```python
import cmath
import math
import matplotlib.pyplot as plt

z = 4 + 3j
a = math.pi / 2

r = cmath.rect(1, a)
nz = z * r

print("before:", z)
print("after:", nz)

x1 = [0, z.real]
y1 = [0, z.imag]

x2 = [0, nz.real]
y2 = [0, nz.imag]

plt.plot(x1, y1)
plt.plot(x2, y2)
plt.show()
```

Computation of DFT using complex numbers

```python
import numpy as np

x = np.array([1.0, 2.0, 3.0, 4.0])
n = len(x)

X = np.zeros(n, dtype=complex)

for k in range(n):
    s = 0j
    for i in range(n):
        s += x[i] * np.exp(-1j * 2 * np.pi * k * i / n)
    X[k] = s

print(X)
```

FFT of given function in tabulated or mathematical form

```python
import numpy as np

def fft(x):
    n = len(x)
    if n == 1:
        return x

    h = n // 2
    w = np.exp(-2j * np.pi * np.arange(h) / n)

    a = x[:h] + x[h:]
    b = (x[:h] - x[h:]) * w

    y1 = fft(a)
    y2 = fft(b)

    ans = np.zeros(n, dtype=complex)
    ans[0::2] = y1
    ans[1::2] = y2

    return ans

f = np.array([1.0, 2.0, 3.0, 4.0, 1.0, 2.0, 3.0, 4.0])
ans = fft(f)

print(ans)
```

Shooting Method (for non lenear equation)


```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y, z):
    return y**2 + x

a = 0
b = 1
ya = 0
yb = 1
n = 100
h = (b - a) / n

def run(m):
    x = a
    y = ya
    z = m
    for i in range(n):
        nz = z + h * f(x, y, z)
        ny = y + h * z
        z = nz
        y = ny
        x = x + h
    return y

m0 = 0.0
m1 = 1.0

for i in range(20):
    e0 = run(m0) - yb
    e1 = run(m1) - yb

    if abs(e1 - e0) < 1e-12:
        break

    m2 = m1 - e1 * (m1 - m0) / (e1 - e0)
    m0 = m1
    m1 = m2

x = a
y = ya
z = m1

xl = [x]
yl = [y]

for i in range(n):
    nz = z + h * f(x, y, z)
    ny = y + h * z
    z = nz
    y = ny
    x = x + h
    xl.append(x)
    yl.append(y)

plt.plot(xl, yl)
plt.show()
```

Shooting Method (for IVP)

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y, z):
    return y**2 + x

a = 0
b = 1
ya = 0
yb = 1
n = 100
h = (b - a) / n

def rk4(m):
    x = a
    y = ya
    z = m
    for i in range(n):
        k1 = h * z
        l1 = h * f(x, y, z)

        k2 = h * (z + l1/2)
        l2 = h * f(x + h/2, y + k1/2, z + l1/2)

        k3 = h * (z + l2/2)
        l3 = h * f(x + h/2, y + k2/2, z + l2/2)

        k4 = h * (z + l3)
        l4 = h * f(x + h, y + k3, z + l3)

        y = y + (k1 + 2*k2 + 2*k3 + k4) / 6
        z = z + (l1 + 2*l2 + 2*l3 + l4) / 6
        x = x + h
    return y

m0 = 0.0
m1 = 1.0

for i in range(20):
    e0 = rk4(m0) - yb
    e1 = rk4(m1) - yb

    if abs(e1 - e0) < 1e-12:
        break

    m2 = m1 - e1 * (m1 - m0) / (e1 - e0)
    m0 = m1
    m1 = m2

x = a
y = ya
z = m1

xl = [x]
yl = [y]

for i in range(n):
    k1 = h * z
    l1 = h * f(x, y, z)

    k2 = h * (z + l1/2)
    l2 = h * f(x + h/2, y + k1/2, z + l1/2)

    k3 = h * (z + l2/2)
    l3 = h * f(x + h/2, y + k2/2, z + l2/2)

    k4 = h * (z + l3)
    l4 = h * f(x + h, y + k3, z + l3)

    y = y + (k1 + 2*k2 + 2*k3 + k4) / 6
    z = z + (l1 + 2*l2 + 2*l3 + l4) / 6
    x = x + h

    xl.append(x)
    yl.append(y)

plt.plot(xl, yl)
plt.show()
```
